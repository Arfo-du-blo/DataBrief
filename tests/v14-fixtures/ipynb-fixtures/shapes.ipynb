```python
import pandas as pd
sales = pd.DataFrame({'region': ['North', 'North', 'South', 'South'], 'year': [2023, 2024, 2023, 2024],
                      'amount': [10.5, 12.0, 7.25, 9.0], 'qty': [3, 4, 2, 5]})
by_region = sales.set_index('region')
by_region
```

```python
pivot = sales.pivot_table(index='region', columns='year', values=['amount', 'qty'], aggfunc='sum')
pivot
```

```python
two_levels = sales.set_index(['region', 'year'])
two_levels
```

```python
wide = pd.DataFrame([[i * j for j in range(30)] for i in range(3)], columns=[f'c{j}' for j in range(30)])
wide
```
